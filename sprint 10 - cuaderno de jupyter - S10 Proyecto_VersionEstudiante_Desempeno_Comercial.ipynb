#### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
Especifica qué KPIs incluirás y por qué: (3-4 KPIs)
- **Ingresos 2024**: punto de partida para comparar la evolución año contra año.
- **Ingresos 2025**: permite comparar directamente contra 2024 sin depender del filtro de año.
- **Variación % de ingresos (2024 vs 2025)**: calculado como `(Ingresos 2025 - Ingresos 2024) / Ingresos 2024`, es el indicador más directo de si el negocio creció o cayó.
- **Margen de ganancia (%)**: calculado como `(Ingresos - Costo) / Ingresos`, muestra si el crecimiento en ingresos se traduce en rentabilidad, no solo en volumen.

Explica qué gráfico usarás para cada pregunta y justifica la elección.

- **Gráfico de líneas** (de tiempo): Muestra `Ingresos` por mes según `Fecha_Pedido`, con 2024 y 2025 diferenciados por color (dos líneas). Justificación: permite ver si la evolución fue constante, estacional, o si hubo un punto específico de caída/subida, algo que los KPIs de tarjeta por sí solos no muestran.
- **Gráfico de barras horizontales** (de comparación geográfica): Muestra `Ingresos` por `País`. Justificación: permite ver de un vistazo qué países generan más ingresos, y con el filtro de año se puede comparar 2024 vs 2025 sin saturar el gráfico.
- **Gráfico de barras verticales** (de comparación clientes/productos): Muestra `Ingresos` por `Categoría_Producto`. Justificación: permite identificar qué categorías de producto generan más ingresos, y con el filtro de año se puede ver cómo cambió esa distribución entre 2024 y 2025.
- **Gráfico de anillo (de rentabilidad):**  Compara Costo vs Ganancia (calculada como Ingresos - Costo). Justificación: da soporte visual al KPI de Margen de ganancia (%), mostrando de forma proporcional cuánto del ingreso se destina a costos y cuánto queda como ganancia real. Se descartó comparar Ingresos vs Costo directamente porque no son partes de un mismo total (el ingreso ya incluye el costo dentro de sí), lo cual generaba un porcentaje engañoso en el gráfico.

Descripción de la jerarquía visual:

- **Título**: "Desempeño Comercial 2024-2025 – Andes Retail Group", en la parte superior izquierda del dashboard, es lo primero que identifica al usuario de qué trata la vista.
- **Segmentadores (`Año`, `País`)**: ubicados en la esquina superior derecha, al mismo nivel del título pero con menor peso visual, ya que son controles de filtro, no protagonistas de la historia.
- **Fila de KPIs**: justo debajo del título, con las 4 tarjetas (`Ingresos 2024`, `Ingresos 2025`, `Variación %`, `Margen de ganancia %`) en una sola fila horizontal. Es lo segundo que ve el usuario, da el resumen ejecutivo en números antes de entrar a los gráficos.
- **Primera fila de gráficos**: el gráfico de líneas (`Ingresos` por Mes y Año) a la izquierda, ocupando más ancho por ser el que responde la pregunta principal de la vista (evolución del ingreso); a su derecha, el gráfico de barras horizontales por `País`, de menor ancho pero con la misma altura.
- **Segunda fila de gráficos**: el gráfico de barras verticales por `Categoría_Producto` a la izquierda, y el gráfico de anillo (Costo vs Ganancia) a la derecha, ambos como apoyo visual de menor jerarquía que la fila superior.

Filtro(s)

- **Año** (`Fecha_Pedido`): filtro principal de la vista, permite alternar entre ver 2024, 2025, o ambos años juntos en todos los gráficos y KPIs a la vez.
- **País**: permite explorar el comportamiento de un país específico sin salir de la Vista Overview.

### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):**  
Andes Retail Group opera en Perú, Chile y Colombia, y en 2024 generó $2.86 millones en ingresos con un margen de ganancia saludable del 35.1%.

**C (Complicación):**  
En 2025 los ingresos cayeron a $2.67 millones, una variación de **-6.8%** respecto a 2024. El patrón mensual es similar entre ambos años (fuerte en los extremos del año, con un valle a mitad de año), pero 2025 se ubica consistentemente por debajo de 2024 en la mayoría de los meses.

**Q (Pregunta):**  
¿Es una caída generalizada del negocio, o se concentra en focos específicos?

**A (Respuesta):**  
No es generalizada. Perú se mantiene como el mercado líder ($2.2 mill.), seguido de Chile ($2.0 mill.) y Colombia ($1.3 mill.), y las categorías de producto están relativamente equilibradas entre sí (~1.3-1.4 mill. cada una). Esto indica que la baja de -6.8% responde a factores puntuales —estacionalidad, región o segmento de cliente— que se investigan en la Vista Detalle, y no a un deterioro uniforme de la operación.

---

#### 🔎 Vista Detalle

**S (Situación):**  
Para explicar la caída de -6.8% vista en el Overview, se analizó el desempeño por estación, región, segmento de cliente y categoría de producto.

**C (Complicación):**  
Se identificaron cuatro focos de deterioro: la estacionalidad es fuerte, con Verano generando 1.22 mill. en 2024 (cayendo a 1.02 mill. en 2025) frente a apenas 0.33 mill. en Invierno; la región Sur tuvo la mayor caída año contra año (0.95 mill. → 0.83 mill.), mientras Norte se mantuvo casi estable (0.94 → 0.95 mill.); el segmento Premium fue el que más cayó (1.40 mill. → 1.20 mill.), mientras que Estándar se mantuvo estable y Económico incluso subió levemente (0.24 → 0.26 mill.); y a nivel categoría, Electrónica en Venta Alta tuvo la peor caída de toda la tabla (-15.5%), seguida de Ropa en Venta Alta (-12.2%).

**Q (Pregunta):**  
¿Estos focos explican la caída total de ingresos, y qué se puede accionar al respecto?

**A (Respuesta):**  
Sí, en conjunto explican la mayor parte de la caída: la fuerte estacionalidad concentra el riesgo en pocos meses del año, la región Sur y el segmento Premium son los principales puntos de fuga de ingresos, y Electrónica/Ropa en el nivel de Venta Alta son las categorías con mayor deterioro. 

Se recomienda reforzar campañas en temporada baja (Invierno), investigar las causas específicas de la caída en la región Sur y en el segmento Premium (posible fuga a competencia o cambio en hábito de compra), y **revisar los productos ofrecidos en Electrónica y Ropa —es decir, si conviene traer productos diferentes, quitar los que no se venden, o cambiar qué modelos/marcas se están priorizando—, además del precio actual de esas categorías**, dado que son las que tienen mayor deterioro en ventas.
